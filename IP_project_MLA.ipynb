```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
from IPython.display import display, HTML, clear_output

# ==========================================
# 0. TEAM WELCOME & VISUAL CREDITS BANNER
# ==========================================
def show_team_welcome_screen():
    display(HTML("""
    <div style="background: linear-gradient(135deg, #0f2027 0%, #203a43 50%, #2c5364 100%);
                color: white; padding: 28px; border-radius: 14px; font-family: 'Segoe UI', sans-serif;
                box-shadow: 0 8px 20px rgba(0,0,0,0.35); margin-bottom: 18px; text-align: center;
                border: 2px solid #00d2ff;">

        <div style="background: rgba(0, 210, 255, 0.18); display: inline-block; padding: 6px 18px;
                    border-radius: 20px; font-size: 13px; font-weight: bold; letter-spacing: 1.5px;
                    color: #00d2ff; margin-bottom: 10px; text-transform: uppercase;">
            🎓 CBSE Class 12 Informatics Practices (IP) Project
        </div>

        <h1 style="margin: 8px 0; font-size: 28px; letter-spacing: 1px; color: #ffffff;">
            📚 SMART LIBRARY MANAGEMENT SYSTEM
        </h1>
        <p style="margin: 4px 0 22px 0; font-size: 14px; color: #cfd8dc;">
            Powered by <b>Python</b> • <b>SQLite3 Relational Database</b> • <b>Pandas DataFrames</b> • <b>Matplotlib Analytics</b>
        </p>

        <div style="background: rgba(255,255,255,0.07); padding: 18px; border-radius: 12px; margin-top: 10px;">
            <h3 style="margin: 0 0 15px 0; font-size: 16px; color: #ffd700; letter-spacing: 1px; text-transform: uppercase;">
                🌟 Project Development Team 🌟
            </h3>

            <div style="display: flex; justify-content: center; gap: 18px; flex-wrap: wrap;">

                <!-- Team Member 1: Monika -->
                <div style="background: linear-gradient(145deg, #1e3c72, #2a5298); padding: 16px 22px;
                            border-radius: 10px; min-width: 180px; border-top: 4px solid #00d2ff;
                            box-shadow: 0 4px 10px rgba(0,0,0,0.25);">
                    <div style="font-size: 28px; margin-bottom: 6px;">👩‍💻</div>
                    <div style="font-size: 18px; font-weight: bold; color: #ffffff;">MONIKA</div>
                    <div style="font-size: 12px; color: #a8d8ff; margin-top: 4px;">Database & SQL Architect</div>
                    <div style="font-size: 11px; color: #e0e0e0; margin-top: 6px; background: rgba(0,0,0,0.25); padding: 3px 8px; border-radius: 4px;">
                        Schema & Queries
                    </div>
                </div>

                <!-- Team Member 2: Ashish -->
                <div style="background: linear-gradient(145deg, #1e3c72, #2a5298); padding: 16px 22px;
                            border-radius: 10px; min-width: 180px; border-top: 4px solid #ffd700;
                            box-shadow: 0 4px 10px rgba(0,0,0,0.25);">
                    <div style="font-size: 28px; margin-bottom: 6px;">👨‍💻</div>
                    <div style="font-size: 18px; font-weight: bold; color: #ffffff;">ASHISH</div>
                    <div style="font-size: 12px; color: #ffe57f; margin-top: 4px;">Core Python & Logic Lead</div>
                    <div style="font-size: 11px; color: #e0e0e0; margin-top: 6px; background: rgba(0,0,0,0.25); padding: 3px 8px; border-radius: 4px;">
                        Transactions & Fine Engine
                    </div>
                </div>

                <!-- Team Member 3: Ajit -->
                <div style="background: linear-gradient(145deg, #1e3c72, #2a5298); padding: 16px 22px;
                            border-radius: 10px; min-width: 180px; border-top: 4px solid #00e676;
                            box-shadow: 0 4px 10px rgba(0,0,0,0.25);">
                    <div style="font-size: 28px; margin-bottom: 6px;">📊</div>
                    <div style="font-size: 18px; font-weight: bold; color: #ffffff;">AJIT</div>
                    <div style="font-size: 12px; color: #b9f6ca; margin-top: 4px;">Data Analytics & UI Lead</div>
                    <div style="font-size: 11px; color: #e0e0e0; margin-top: 6px; background: rgba(0,0,0,0.25); padding: 3px 8px; border-radius: 4px;">
                        Pandas & Matplotlib Charts
                    </div>
                </div>

            </div>
        </div>

        <div style="margin-top: 16px; font-size: 12px; color: #b0bec5;">
            ✅ Status: <b>SQL Database Connected</b> &nbsp;|&nbsp; ✅ Modules: <b>All Active</b> &nbsp;|&nbsp; 📅 Session: <b>2026-27</b>
        </div>
    </div>
    """))

# ==========================================
# 1. DATABASE INITIALIZATION & SQL SETUP
# ==========================================
def init_db():
    conn = sqlite3.connect('library_system.db')
    cursor = conn.cursor()

    # Create Books Table
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS books (
            book_id INTEGER PRIMARY KEY AUTOINCREMENT,
            title TEXT NOT NULL,
            author TEXT NOT NULL,
            category TEXT NOT NULL,
            total_copies INTEGER NOT NULL,
            available_copies INTEGER NOT NULL
        )
    ''')

    # Create Members Table
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS members (
            member_id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL,
            class_sec TEXT NOT NULL,
            phone TEXT NOT NULL,
            join_date TEXT NOT NULL
        )
    ''')

    # Create Transactions Table (Issue / Return)
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS transactions (
            trans_id INTEGER PRIMARY KEY AUTOINCREMENT,
            book_id INTEGER,
            member_id INTEGER,
            issue_date TEXT NOT NULL,
            due_date TEXT NOT NULL,
            return_date TEXT,
            fine REAL DEFAULT 0.0,
            status TEXT DEFAULT 'Issued',
            FOREIGN KEY (book_id) REFERENCES books(book_id),
            FOREIGN KEY (member_id) REFERENCES members(member_id)
        )
    ''')

    # Seed Pre-Populated Data if Database is Empty
    cursor.execute("SELECT COUNT(*) FROM books")
    if cursor.fetchone()[0] == 0:
        sample_books = [
            ('Informatics Practices Class 12', 'Sumita Arora', 'Computer Science', 10, 8),
            ('Python for Data Analysis', 'Wes McKinney', 'Computer Science', 5, 5),
            ('Concepts of Physics Vol 1', 'H.C. Verma', 'Physics', 8, 7),
            ('Organic Chemistry', 'Morrison & Boyd', 'Chemistry', 6, 6),
            ('Higher Engineering Mathematics', 'B.S. Grewal', 'Mathematics', 7, 6),
            ('Artificial Intelligence: A Modern Approach', 'Stuart Russell', 'AI & ML', 4, 3),
            ('Wings of Fire', 'A.P.J. Abdul Kalam', 'Biography', 6, 6)
        ]
        cursor.executemany('''
            INSERT INTO books (title, author, category, total_copies, available_copies)
            VALUES (?, ?, ?, ?, ?)
        ''', sample_books)

        sample_members = [
            ('Monika', '12-A', '9876543210', '2026-04-01'),
            ('Ashish', '12-A', '9876543211', '2026-04-02'),
            ('Ajit', '12-A', '9876543212', '2026-04-03'),
            ('Jignesh Verma', '12-B', '9876543213', '2026-04-05'),
            ('Radhika Singh', '12-B', '9876543214', '2026-04-06')
        ]
        cursor.executemany('''
            INSERT INTO members (name, class_sec, phone, join_date)
            VALUES (?, ?, ?, ?)
        ''', sample_members)

        # Sample Transactions (Including overdue and returned cases)
        today = datetime.now()
        sample_trans = [
            (1, 1, (today - timedelta(days=5)).strftime('%Y-%m-%d'), (today + timedelta(days=9)).strftime('%Y-%m-%d'), None, 0.0, 'Issued'),
            (1, 2, (today - timedelta(days=20)).strftime('%Y-%m-%d'), (today - timedelta(days=6)).strftime('%Y-%m-%d'), None, 0.0, 'Issued'),
            (3, 3, (today - timedelta(days=3)).strftime('%Y-%m-%d'), (today + timedelta(days=11)).strftime('%Y-%m-%d'), None, 0.0, 'Issued'),
            (5, 4, (today - timedelta(days=10)).strftime('%Y-%m-%d'), (today + timedelta(days=4)).strftime('%Y-%m-%d'), None, 0.0, 'Issued'),
            (6, 5, (today - timedelta(days=2)).strftime('%Y-%m-%d'), (today + timedelta(days=12)).strftime('%Y-%m-%d'), None, 0.0, 'Issued'),
            (2, 1, (today - timedelta(days=25)).strftime('%Y-%m-%d'), (today - timedelta(days=11)).strftime('%Y-%m-%d'), (today - timedelta(days=5)).strftime('%Y-%m-%d'), 30.0, 'Returned')
        ]
        cursor.executemany('''
            INSERT INTO transactions (book_id, member_id, issue_date, due_date, return_date, fine, status)
            VALUES (?, ?, ?, ?, ?, ?, ?)
        ''', sample_trans)

    conn.commit()
    return conn

# ==========================================
# 2. UI HELPER FUNCTIONS FOR GOOGLE COLAB
# ==========================================
def print_banner(title):
    display(HTML(f"""
    <div style="background: linear-gradient(90deg, #1e3c72 0%, #2a5298 100%);
                color: white; padding: 12px 20px; border-radius: 8px;
                font-family: sans-serif; margin: 12px 0; box-shadow: 0 2px 5px rgba(0,0,0,0.2);">
        <h3 style="margin: 0; font-size: 18px;">📚 {title}</h3>
    </div>
    """))

def print_msg(text, msg_type="success"):
    colors = {
        "success": ("#d4edda", "#155724", "✅"),
        "error": ("#f8d7da", "#721c24", "❌"),
        "info": ("#d1ecf1", "#0c5460", "ℹ️")
    }
    bg, fg, icon = colors.get(msg_type, colors["info"])
    display(HTML(f"""
    <div style="background-color: {bg}; color: {fg}; padding: 10px 15px;
                border-radius: 6px; margin: 8px 0; font-family: sans-serif; font-weight: bold;">
        {icon} {text}
    </div>
    """))

def show_sql_table(conn, query, params=()):
    df = pd.read_sql_query(query, conn, params=params)
    if df.empty:
        print_msg("No records found matching your query.", "info")
    else:
        display(df.style.set_properties(**{
            'text-align': 'left',
            'padding': '8px',
            'border': '1px solid #ddd'
        }).set_table_styles([
            {'selector': 'th', 'props': [('background-color', '#2a5298'), ('color', 'white'), ('text-align', 'left'), ('padding', '8px')]}
        ]).hide(axis='index'))

# ==========================================
# 3. CORE LIBRARY MANAGEMENT MODULES
# ==========================================
def view_all_books(conn):
    print_banner("COMPLETE BOOK INVENTORY")
    query = """
        SELECT book_id AS 'Book ID', title AS 'Title', author AS 'Author',
               category AS 'Category', total_copies AS 'Total Copies',
               available_copies AS 'Available Copies'
        FROM books
    """
    show_sql_table(conn, query)

def add_new_book(conn):
    print_banner("ADD NEW BOOK TO INVENTORY")
    try:
        title = input("Enter Book Title: ").strip()
        author = input("Enter Author Name: ").strip()
        category = input("Enter Category (e.g., Physics, CS, AI & ML): ").strip()
        copies = int(input("Enter Total Number of Copies: "))

        if not title or not author or copies <= 0:
            print_msg("Invalid details! Title/Author cannot be empty and copies must be greater than 0.", "error")
            return

        cursor = conn.cursor()
        cursor.execute('''
            INSERT INTO books (title, author, category, total_copies, available_copies)
            VALUES (?, ?, ?, ?, ?)
        ''', (title, author, category, copies, copies))
        conn.commit()
        print_msg(f"Book '{title}' successfully added to the database!", "success")
    except ValueError:
        print_msg("Error: Copies must be a valid integer.", "error")

def register_member(conn):
    print_banner("REGISTER NEW LIBRARY MEMBER")
    name = input("Enter Student/Member Name: ").strip()
    class_sec = input("Enter Class & Section (e.g., 12-A): ").strip()
    phone = input("Enter 10-Digit Phone Number: ").strip()
    join_date = datetime.now().strftime('%Y-%m-%d')

    if not name or not class_sec:
        print_msg("Name and Class/Section are mandatory!", "error")
        return

    cursor = conn.cursor()
    cursor.execute('''
        INSERT INTO members (name, class_sec, phone, join_date)
        VALUES (?, ?, ?, ?)
    ''', (name, class_sec, phone, join_date))
    conn.commit()
    print_msg(f"Member '{name}' registered successfully! Assigned Member ID: {cursor.lastrowid}", "success")

def view_members(conn):
    print_banner("REGISTERED LIBRARY MEMBERS")
    query = """
        SELECT member_id AS 'Member ID', name AS 'Student Name',
               class_sec AS 'Class/Sec', phone AS 'Contact No', join_date AS 'Joined Date'
        FROM members
    """
    show_sql_table(conn, query)

def issue_book(conn):
    print_banner("ISSUE A BOOK")
    view_all_books(conn)
    try:
        book_id = int(input("\nEnter Book ID to Issue: "))
        member_id = int(input("Enter Member ID: "))

        cursor = conn.cursor()
        cursor.execute("SELECT title, available_copies FROM books WHERE book_id = ?", (book_id,))
        book = cursor.fetchone()

        if not book:
            print_msg("Book ID does not exist!", "error")
            return
        if book[1] <= 0:
            print_msg(f"Sorry, '{book[0]}' is currently out of stock!", "error")
            return

        cursor.execute("SELECT name FROM members WHERE member_id = ?", (member_id,))
        member = cursor.fetchone()
        if not member:
            print_msg("Member ID does not exist!", "error")
            return

        issue_date = datetime.now().strftime('%Y-%m-%d')
        due_date = (datetime.now() + timedelta(days=14)).strftime('%Y-%m-%d')

        cursor.execute('''
            INSERT INTO transactions (book_id, member_id, issue_date, due_date, status)
            VALUES (?, ?, ?, ?, 'Issued')
        ''', (book_id, member_id, issue_date, due_date))

        cursor.execute('''
            UPDATE books SET available_copies = available_copies - 1 WHERE book_id = ?
        ''', (book_id,))

        conn.commit()
        print_msg(f"Book '{book[0]}' issued to {member[0]}! Due Date: {due_date} (14 Days)", "success")
    except ValueError:
        print_msg("Invalid input! Please enter numeric IDs.", "error")

def return_book(conn):
    print_banner("RETURN A BOOK & CALCULATE FINE")
    query = """
        SELECT t.trans_id AS 'Trans ID', b.title AS 'Book Title', m.name AS 'Member Name',
               t.issue_date AS 'Issue Date', t.due_date AS 'Due Date', t.status AS 'Status'
        FROM transactions t
        JOIN books b ON t.book_id = b.book_id
        JOIN members m ON t.member_id = m.member_id
        WHERE t.status = 'Issued'
    """
    show_sql_table(conn, query)

    try:
        trans_id = int(input("\nEnter Transaction ID (Trans ID) to Return: "))
        cursor = conn.cursor()
        cursor.execute("""
            SELECT book_id, due_date FROM transactions
            WHERE trans_id = ? AND status = 'Issued'
        """, (trans_id,))
        record = cursor.fetchone()

        if not record:
            print_msg("Invalid Transaction ID or Book already returned!", "error")
            return

        book_id, due_date_str = record
        due_date = datetime.strptime(due_date_str, '%Y-%m-%d')
        today = datetime.now()
        return_date_str = today.strftime('%Y-%m-%d')

        overdue_days = (today - due_date).days
        fine = max(0, overdue_days * 5.0)

        cursor.execute('''
            UPDATE transactions
            SET return_date = ?, fine = ?, status = 'Returned'
            WHERE trans_id = ?
        ''', (return_date_str, fine, trans_id))

        cursor.execute('''
            UPDATE books SET available_copies = available_copies + 1 WHERE book_id = ?
        ''', (book_id,))

        conn.commit()
        if fine > 0:
            print_msg(f"Book returned successfully! Late by {overdue_days} days. Overdue Fine: ₹{fine:.2f}", "info")
        else:
            print_msg("Book returned on time! No fine charged (₹0.00).", "success")
    except ValueError:
        print_msg("Invalid input! Transaction ID must be a number.", "error")

def search_books(conn):
    print_banner("SEARCH BOOKS BY TITLE, AUTHOR, OR CATEGORY")
    keyword = input("Enter search keyword: ").strip()
    query = """
        SELECT book_id AS 'Book ID', title AS 'Title', author AS 'Author',
               category AS 'Category', total_copies AS 'Total', available_copies AS 'Available'
        FROM books
        WHERE title LIKE ? OR author LIKE ? OR category LIKE ?
    """
    wildcard = f"%{keyword}%"
    show_sql_table(conn, query, (wildcard, wildcard, wildcard))

def view_transactions(conn):
    print_banner("ALL TRANSACTION LOGS (SQL JOIN REPORT)")
    query = """
        SELECT t.trans_id AS 'Trans ID', b.title AS 'Book Title', m.name AS 'Issued To',
               m.class_sec AS 'Class', t.issue_date AS 'Issue Date', t.due_date AS 'Due Date',
               IFNULL(t.return_date, 'Not Returned') AS 'Return Date',
               '₹' || printf('%.2f', t.fine) AS 'Fine', t.status AS 'Status'
        FROM transactions t
        JOIN books b ON t.book_id = b.book_id
        JOIN members m ON t.member_id = m.member_id
        ORDER BY t.trans_id DESC
    """
    show_sql_table(conn, query)

def delete_book(conn):
    print_banner("DELETE A BOOK RECORD")
    view_all_books(conn)
    try:
        book_id = int(input("\nEnter Book ID to Delete permanently: "))
        cursor = conn.cursor()
        cursor.execute("SELECT title FROM books WHERE book_id = ?", (book_id,))
        row = cursor.fetchone()
        if not row:
            print_msg("Book ID not found!", "error")
            return
        cursor.execute("DELETE FROM books WHERE book_id = ?", (book_id,))
        conn.commit()
        print_msg(f"Book '{row[0]}' (ID: {book_id}) deleted from database.", "success")
    except ValueError:
        print_msg("Invalid Book ID!", "error")

# ==========================================
# 4. DATA VISUALIZATION MODULE (MATPLOTLIB)
# ==========================================
def show_analytics(conn):
    print_banner("LIBRARY ANALYTICS & DATA VISUALIZATION (MATPLOTLIB)")

    df_cat = pd.read_sql_query("""
        SELECT category, SUM(total_copies) as total, SUM(available_copies) as available
        FROM books GROUP BY category
    """, conn)

    df_status = pd.read_sql_query("""
        SELECT status, COUNT(*) as count FROM transactions GROUP BY status
    """, conn)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    x = range(len(df_cat['category']))
    width = 0.35
    ax1.bar([i - width/2 for i in x], df_cat['total'], width=width, label='Total Copies', color='#1e3c72')
    ax1.bar([i + width/2 for i in x], df_cat['available'], width=width, label='Available Copies', color='#00d2ff')
    ax1.set_title('Book Stock by Category', fontsize=13, fontweight='bold')
    ax1.set_xticks(list(x))
    ax1.set_xticklabels(df_cat['category'], rotation=25, ha='right')
    ax1.set_ylabel('Number of Copies')
    ax1.legend()
    ax1.grid(axis='y', linestyle='--', alpha=0.5)

    if not df_status.empty:
        ax2.pie(df_status['count'], labels=df_status['status'], autopct='%1.1f%%',
                startangle=140, colors=['#ff9966', '#28a745'], explode=[0.05]*len(df_status))
        ax2.set_title('Transactions Breakdown (Issued vs Returned)', fontsize=13, fontweight='bold')

    plt.tight_layout()
    plt.show()

# ==========================================
# 5. MAIN INTERACTIVE COLAB MENU
# ==========================================
def main_menu():
    conn = init_db()
    first_launch = True

    while True:
        if first_launch:
            show_team_welcome_screen()
            first_launch = False
        else:
            print_banner("SMART LIBRARY MANAGEMENT SYSTEM | TEAM: MONIKA • ASHISH • AJIT")

        print("  [1] 📖 View All Books Inventory")
        print("  [2] ➕ Add a New Book")
        print("  [3] 🔍 Search Books (By Title / Author / Category)")
        print("  [4] 👥 View Registered Members")
        print("  [5] 👤 Register New Student Member")
        print("  [6] 📤 Issue a Book to Member")
        print("  [7] 📥 Return a Book & Auto-Calculate Fine")
        print("  [8] 📋 View Complete Transaction History (SQL JOIN)")
        print("  [9] 📊 Generate Visual Analytics Charts (Matplotlib)")
        print("  [10] 🗑️ Delete a Book Record")
        print("  [11] 🌟 View Team Presentation Banner (Monika, Ashish & Ajit)")
        print("  [0] 🚪 Exit System")
        print("=" * 68)

        choice = input("Select an option (0-11): ").strip()
        clear_output(wait=True)

        if choice == '1':
            view_all_books(conn)
        elif choice == '2':
            add_new_book(conn)
        elif choice == '3':
            search_books(conn)
        elif choice == '4':
            view_members(conn)
        elif choice == '5':
            register_member(conn)
        elif choice == '6':
            issue_book(conn)
        elif choice == '7':
            return_book(conn)
        elif choice == '8':
            view_transactions(conn)
        elif choice == '9':
            show_analytics(conn)
        elif choice == '10':
            delete_book(conn)
        elif choice == '11':
            show_team_welcome_screen()
        elif choice == '0':
            show_team_welcome_screen()
            print_msg("Thank you! Project presented by Monika, Ashish & Ajit. Database saved safely.", "info")
            conn.close()
            break
        else:
            print_msg("Invalid choice! Please select a valid option between 0 and 11.", "error")

        input("\nPress [Enter] to return to the Main Menu...")
        clear_output(wait=True)

if __name__ == "__main__":
    main_menu()
```
